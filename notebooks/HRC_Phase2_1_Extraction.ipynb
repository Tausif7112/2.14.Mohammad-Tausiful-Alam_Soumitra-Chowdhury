```python
# Verify the final extracted skeleton and label files (shape, dtype, and per-class sample counts)

import cv2
import numpy as np
import mediapipe as mp
from mediapipe.tasks import python as mp_python
from mediapipe.tasks.python import vision as mp_vision
import urllib.request
import os

print(f"MediaPipe version: {mp.__version__} ")

model_path = '/content/pose_landmarker_lite.task'
if not os.path.exists(model_path):
    print("Downloading pose landmarker model...")
    urllib.request.urlretrieve(
        'https://storage.googleapis.com/mediapipe-models/pose_landmarker/pose_landmarker_lite/float16/latest/pose_landmarker_lite.task',
        model_path
    )
    print(f"Model downloaded: {os.path.getsize(model_path)/1024/1024:.1f} MB ")
else:
    print("Model already downloaded ")

base_options = mp_python.BaseOptions(model_asset_path=model_path)
options = mp_vision.PoseLandmarkerOptions(
    base_options=base_options,
    running_mode=mp_vision.RunningMode.VIDEO,
    num_poses=1,
    min_pose_detection_confidence=0.5,
    min_pose_presence_confidence=0.5,
    min_tracking_confidence=0.5
)

def extract_skeleton(video_path, landmarker):
    """
    Extracts MediaPipe Pose landmarks from a video.
    Returns numpy array shape (3, 150, 33):
        3  = x, y, z coordinates
        150 = fixed frames (center-crop if longer, zero-pad if shorter)
        33 = MediaPipe Pose landmarks
    Frames with no detected pose are filled with zeros.
    """
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        raise ValueError(f"Cannot open: {video_path}")

    fps = cap.get(cv2.CAP_PROP_FPS)
    if fps <= 0:
        fps = 30.0

    frames_data = []
    frame_idx = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=frame_rgb)
        timestamp_ms = int((frame_idx / fps) * 1000)

        result = landmarker.detect_for_video(mp_image, timestamp_ms)

        if result.pose_landmarks and len(result.pose_landmarks) > 0:
            landmarks = np.array(
                [[lm.x, lm.y, lm.z] for lm in result.pose_landmarks[0]],
                dtype=np.float32
            )
        else:
            landmarks = np.zeros((33, 3), dtype=np.float32)

        frames_data.append(landmarks)
        frame_idx += 1

    cap.release()

    if len(frames_data) == 0:
        frames_data = [np.zeros((33, 3), dtype=np.float32)]

    arr = np.array(frames_data, dtype=np.float32)
    T = arr.shape[0]
    target_T = 150

    if T > target_T:
        start = (T - target_T) // 2
        arr = arr[start:start + target_T]
    elif T < target_T:
        pad = np.zeros((target_T - T, 33, 3), dtype=np.float32)
        arr = np.concatenate([arr, pad], axis=0)

    return np.transpose(arr, (2, 0, 1))

test_video = '/content/HRI30_videos/videos/DeliverObject/v_DeliverObject_g01_c01.avi'
print(f"\nTesting on: {os.path.basename(test_video)}")

with mp_vision.PoseLandmarker.create_from_options(options) as landmarker:
    result = extract_skeleton(test_video, landmarker)

print(f"Output shape: {result.shape}")
if result.shape == (3, 150, 33):
    print("Shape correct  — ready for full extraction")
else:
    print(" Wrong shape — stop and report error")
```

```python
# Full extraction
import cv2
import numpy as np
import mediapipe as mp
from mediapipe.tasks import python as mp_python
from mediapipe.tasks.python import vision as mp_vision
import os, time
from pathlib import Path

videos_base = Path('/content/HRI30_videos/videos')
output_dir  = Path('/content/drive/MyDrive/HRC_Research/datasets/HRI30/extracted_skeletons')
output_dir.mkdir(parents=True, exist_ok=True)
model_path  = '/content/pose_landmarker_lite.task'

classind_path = '/content/HRI30_annotations/annotations/classInd.txt'
class_to_label = {}
with open(classind_path, 'r') as f:
    for line in f:
        line = line.strip()
        if line:
            idx, name = line.split(' ', 1)
            class_to_label[name] = int(idx) - 1
print(f"Class map: {len(class_to_label)} classes ")

video_list = []
for class_name, label in sorted(class_to_label.items(), key=lambda x: x[1]):
    class_dir = videos_base / class_name
    if not class_dir.exists():
        continue
    for fname in sorted(os.listdir(class_dir)):
        if fname.endswith('.avi'):
            video_list.append((str(class_dir / fname), label))
print(f"Total videos queued: {len(video_list)} ")

base_options = mp_python.BaseOptions(model_asset_path=model_path)
options = mp_vision.PoseLandmarkerOptions(
    base_options=base_options,
    running_mode=mp_vision.RunningMode.VIDEO,
    num_poses=1,
    min_pose_detection_confidence=0.5,
    min_pose_presence_confidence=0.5,
    min_tracking_confidence=0.5
)

def extract_skeleton(video_path, mp_options):
    with mp_vision.PoseLandmarker.create_from_options(mp_options) as landmarker:
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            raise ValueError(f"Cannot open: {video_path}")

        fps = cap.get(cv2.CAP_PROP_FPS)
        if fps <= 0:
            fps = 30.0

        frames_data = []
        frame_idx = 0

        while True:
            ret, frame = cap.read()
            if not ret:
                break

            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=frame_rgb)

            timestamp_ms = int((frame_idx / fps) * 1000)

            result = landmarker.detect_for_video(mp_image, timestamp_ms)

            if result.pose_landmarks and len(result.pose_landmarks) > 0:
                landmarks = np.array(
                    [[lm.x, lm.y, lm.z] for lm in result.pose_landmarks[0]],
                    dtype=np.float32)
            else:
                landmarks = np.zeros((33, 3), dtype=np.float32)

            frames_data.append(landmarks)
            frame_idx += 1

        cap.release()

        if len(frames_data) == 0:
            frames_data = [np.zeros((33, 3), dtype=np.float32)]

        arr = np.array(frames_data, dtype=np.float32)
        T = arr.shape[0]
        target_T = 150

        if T > target_T:
            start = (T - target_T) // 2
            arr = arr[start:start + target_T]
        elif T < target_T:
            arr = np.concatenate(
                [arr, np.zeros((target_T - T, 33, 3), dtype=np.float32)], axis=0)

        return np.transpose(arr, (2, 0, 1))

all_skeletons   = []
all_labels      = []
failed_videos   = []
batch_skeletons = []
batch_labels    = []
batch_num       = 0
CHECKPOINT_EVERY = 100
start_time = time.time()

print("Starting extraction... This will take 2-3 hours. Do not close tab.")

for i, (video_path, label) in enumerate(video_list):
    try:
        skeleton = extract_skeleton(video_path, options)
        batch_skeletons.append(skeleton)
        batch_labels.append(label)
    except Exception as e:
        print(f"  FAILED [{i+1}]: {os.path.basename(video_path)} — {e}")
        failed_videos.append(video_path)

    if (i + 1) % 50 == 0:
        elapsed = (time.time() - start_time) / 60
        rate = (i + 1) / elapsed if elapsed > 0 else 0
        remaining = (len(video_list) - (i + 1)) / rate if rate > 0 else 0
        print(f"[{i+1}/{len(video_list)}] "
              f"Elapsed: {elapsed:.1f} min | "
              f"ETA: {remaining:.1f} min | "
              f"Failed: {len(failed_videos)}")

    if len(batch_skeletons) >= CHECKPOINT_EVERY:
        batch_num += 1
        ckpt = output_dir / f'checkpoint_batch_{batch_num:04d}.npz'
        np.savez(ckpt,
                 skeletons=np.array(batch_skeletons, dtype=np.float32),
                 labels=np.array(batch_labels, dtype=np.int32))
        print(f"  Checkpoint {batch_num} saved to Drive "
              f"({CHECKPOINT_EVERY} videos)")
        all_skeletons.extend(batch_skeletons)
        all_labels.extend(batch_labels)
        batch_skeletons = []
        batch_labels    = []

if batch_skeletons:
    batch_num += 1
    ckpt = output_dir / f'checkpoint_batch_{batch_num:04d}.npz'
    np.savez(ckpt,
             skeletons=np.array(batch_skeletons, dtype=np.float32),
             labels=np.array(batch_labels, dtype=np.int32))
    print(f" Final checkpoint {batch_num} saved ({len(batch_skeletons)} videos)")
    all_skeletons.extend(batch_skeletons)
    all_labels.extend(batch_labels)

print("\nSaving final .npy files to Drive...")
final_skeletons = np.array(all_skeletons, dtype=np.float32)
final_labels    = np.array(all_labels,    dtype=np.int32)

np.save(output_dir / 'hri30_skeletons.npy', final_skeletons)
np.save(output_dir / 'hri30_labels.npy',    final_labels)

total_min = (time.time() - start_time) / 60
print(f"\nhri30_skeletons.npy shape: {final_skeletons.shape}")
print(f"hri30_labels.npy shape:    {final_labels.shape} ")
print(f"Total failed videos:       {len(failed_videos)}")
print(f"Total time:                {total_min:.1f} minutes")
print(f"\n Phase 2 Step 2.1 COMPLETE")
```
