Copying NTU60 zip to Colab local disk first

```python
import shutil, os

src = '/content/drive/MyDrive/nturgbd_skeletons_s001_to_s017.zip'
dst = '/content/nturgbd_skeletons_s001_to_s017.zip'

print("Copying zip from Drive to Colab local disk...")

shutil.copy2(src, dst)

size = os.path.getsize(dst) / 1024**3
print(f"\nCopy complete")
print(f"Local file size: {size:.2f} GB")
```

Unzipping dataset in colab local environment

```python
import zipfile, os

zip_path = '/content/nturgbd_skeletons_s001_to_s017.zip'
extract_to = '/content/NTU60_raw/'

print("Unzipping NTU60 locally in Colab...")
print("This may take 5-10 minutes...")

os.makedirs(extract_to, exist_ok=True)

with zipfile.ZipFile(zip_path, 'r') as z:
    z.extractall(extract_to)

# Verify
files = []
for root, dirs, filenames in os.walk(extract_to):
    files.extend(filenames)

print(f"\nExtraction complete")
print(f"Total files extracted: {len(files)}")
print(f"Expected: 56,880")
```
